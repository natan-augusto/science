## 2. Theory

### Definition:
A proposition is any sentence that:

- has a subject and a predicate;
- is declarative;
- has one, and only one, of two possible logical values: **true** or **false**;

### Negation:
For every proposition $p$, there is a corresponding negation of $p$ ($\neg p$). Every negation of p has a logical truth value opposite to that of proposition p; therefore, we can conclude that:  

> "The proposition $\neg p$ always has the opposite value of $p$; that is, $\neg p$ is true when p is false, and $\neg p$ is false when $p$ is true."

### Truth-Table:
It is a mathematical tool for showing the possible logical values of a proposition. For $p$, we have:

<div align="center">

| $p$ | $\neg p$ |
| :--- | ---: |
| V | F |
| F | V |
</div>

## 3. Python Application
</div>

```python
import pandas as pd
print("For logical values, enter T (True) or F (False):")
print()
logicalvalue = input(print(r"Enter the truth value of the proposition p: "))

if logicalvalue == "T" or logicalvalue == "t":
    print("If p is true, then ¬p is false")
    truthtable = {
         "p": ["V"],
         "¬p": ["F"]
    }
    df = pd.DataFrame(truthtable)
    print(df)
if logicalvalue == "F" or logicalvalue == "f":
    print("If p is false, then ¬p is true")
    truthtable_2 = {
         "p": ["F"],
         "¬p": ["V"]
    }
    df = pd.DataFrame(truthtable_2)
    print(df)
if logicalvalue != "T" and logicalvalue != "F" and logicalvalue != "t" and logicalvalue != "f":
    print("\033[31mInvalid input. Please enter T or F.\033[0m")
```

# 4. Conclusion
> It can be concluded that for every proposition, there is a counter-proposition called negation, which will always have a logical value opposite to that of the original proposition.
